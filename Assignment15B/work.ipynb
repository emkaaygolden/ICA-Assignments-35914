```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('student_tea_snacks.csv')
df
```

```python
print("Most prefered tea type")
print(df['tea_type'].value_counts().idxmax())
```

```python
print("What is the average number of cups per gender?")
df.groupby('gender')['cups_per_day'].mean().round(2)
```

```python
df['snack_choice'].value_counts().plot(kind='bar')

plt.title('Most Popular Snacks')
plt.xlabel('Snack Choice')
plt.ylabel('Number of Students')

plt.show()
```

```python
df.groupby(['age_group', 'prefers_tea']).size().unstack().plot(kind='bar')

plt.title('Tea Preference by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Students')
plt.show()
```

```python
print("Number of students who dont drink tea: ")
print((df['prefers_tea'] == 'No').sum())
```
