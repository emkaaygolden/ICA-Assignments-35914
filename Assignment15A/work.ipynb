```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ride_hailing_dataset.csv')
df
```

```python
print("Average Fare - by service type")
print(df.groupby('service_type')['fare'].mean().round(2))
```

```python
print("Which City has the most trips?")
print(df['city'].value_counts().idxmax())
```

```python
print("What is the average distance per trip?")
print(df['distance_km'].mean().round(2))
```

```python
print("Trip ratings by gender:")
df.groupby('gender')['trip_rating'].mean().round(2)
```

```python
chart = df.groupby('service_type')['fare'].mean().plot(kind='bar')
plt.title('Average Fare by Service Type')
plt.xlabel('Service Type')
plt.ylabel('Average Fare')

plt.show()

```
