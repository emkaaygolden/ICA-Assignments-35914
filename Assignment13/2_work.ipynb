```python
import pandas as pd
```

```python
df = pd.read_csv("teaching_prices_clean.csv")
```

```python
summary = df.pivot_table(index="item", values="price", aggfunc=["mean", "median"]).round(2)
summary.columns = ["Mean", "Median"]
```

```python
weekly = df.pivot_table(index="item", columns="date", values="price", aggfunc="mean")
```

```python
first = weekly["2026-01-01"]
last = weekly["2026-03-19"]
summary["Pct_Change"] = ((last - first) / first * 100).round(2)
```
