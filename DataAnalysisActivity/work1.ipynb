```python
import pandas as pd
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('pakistani_ecommerce_dataset.csv')
df
```

```python
print (f"Most Popular Category: {df['product_category'].value_counts().idxmax()}")
```

```python
print (f"Average Order Value: {df['order_value'].mean().round(2)}")
```

```python
print (f"City with highest number of orders: {df['city'].value_counts().idxmax()}")
```

```python
print(f"Most Used Payment Method: {df['payment_method'].value_counts().idxmax()}")
```

```python
df['product_category'].value_counts().plot(kind='bar')
plt.xlabel('Product Category')
plt.ylabel("Number of Orders")
plt.title("Orders by Category")
plt.show()
```
