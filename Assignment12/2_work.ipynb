```python
import pandas as pd

df = pd.read_csv('teaching_prices_messy.csv')
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()
df_clean.loc[df_clean['item'] == 'oil b', 'item'] = 'Staple A'
df_clean['price'] = pd.to_numeric(df_clean['price'].astype(str).str.replace(',', ''), errors='coerce')
df_clean['price'] = df_clean['price'].replace(9999.0, float('nan'))

print(df.shape, df_clean.shape)
print(df.isnull().sum(), df_clean.isnull().sum())
print(df.describe(include='all'))
print(df_clean.describe(include='all'))
print(df_clean.groupby(['item', 'city']).size())
```
